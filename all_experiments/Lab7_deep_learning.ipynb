```python
%matplotlib inline
import json
import os
import site
import subprocess
import sys
import zipfile
from pathlib import Path

os.environ.setdefault("KMP_DUPLICATE_LIB_OK", "TRUE")

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.metrics import ConfusionMatrixDisplay, accuracy_score

try:
    import torch
except ModuleNotFoundError:
    print("PyTorch is not installed for", sys.executable, "- installing it once ...")
    pip = [sys.executable, "-m", "pip", "install", "torch"]
    if subprocess.run(pip + ["--index-url", "https://download.pytorch.org/whl/cpu"]).returncode != 0:
        subprocess.run(pip, check=True)
    site.addsitedir(site.getusersitepackages())
    import torch
from torch import nn

if torch.cuda.is_available():
    DEVICE = torch.device("cuda")
elif getattr(torch.backends, "mps", None) is not None and torch.backends.mps.is_available():
    DEVICE = torch.device("mps")
else:
    DEVICE = torch.device("cpu")

torch.manual_seed(42)

PROJECT_ROOT = Path.cwd().parent
DATA = PROJECT_ROOT / "all_datasets"
MODELS = PROJECT_ROOT / "models"
MODELS.mkdir(exist_ok=True)

print("Python", sys.version.split()[0], "at", sys.executable)
print("PyTorch", torch.__version__, "| device:", DEVICE,
      f"({torch.cuda.get_device_name(0)})" if DEVICE.type == "cuda" else "")
```

**Setup.** PyTorch 2.15.0 runs on the `mps` device (Apple GPU). The seed is fixed at 42 so results are reproducible, and `models/` is created next to `all_experiments`.

```python
columns = [f"pixel_{i}" for i in range(64)] + ["label"]
with zipfile.ZipFile(DATA / "optical_recognition_handwritten_digits.zip") as archive:
    members = {Path(name).name: name for name in archive.namelist()}
    train = pd.read_csv(archive.open(members["optdigits.tra"]), header=None, names=columns)
    test = pd.read_csv(archive.open(members["optdigits.tes"]), header=None, names=columns)

def to_tensors(frame):
    X = torch.tensor(frame.drop(columns="label").to_numpy(), dtype=torch.float32)
    X = X.reshape(-1, 1, 8, 8) / 16.0
    y = torch.tensor(frame["label"].to_numpy(), dtype=torch.long)
    return X.to(DEVICE), y.to(DEVICE)

X_dev, y_dev = to_tensors(train)
X_test, y_test = to_tensors(test)

perm = torch.randperm(len(X_dev), generator=torch.Generator().manual_seed(42))
n_val = len(X_dev) // 4
X_val, y_val = X_dev[perm[:n_val]], y_dev[perm[:n_val]]
X_fit, y_fit = X_dev[perm[n_val:]], y_dev[perm[n_val:]]
print("fit:", X_fit.shape, " validation:", X_val.shape, " test:", X_test.shape)

fig, axes = plt.subplots(1, 10, figsize=(10, 1.4))
for ax, image, label in zip(axes, X_fit[:10], y_fit[:10]):
    ax.imshow(image[0].cpu(), cmap="gray_r")
    ax.set_title(int(label), fontsize=9)
    ax.axis("off")
plt.suptitle("Ten training digits as the network sees them", y=1.05)
plt.show()
```

**Part A.** Digits are reshaped to (1, 8, 8) and divided by 16 so values sit in [0, 1]. The split is 2868 fit, 955 validation (one quarter of the training rows) and 1797 official test. The test file is only used at the end.

```python
def make_mlp():
    return nn.Sequential(
        nn.Flatten(),
        nn.Linear(64, 128), nn.ReLU(),
        nn.Linear(128, 10),
    )

mlp = make_mlp().to(DEVICE)
print(mlp)
print("trainable weights:", sum(p.numel() for p in mlp.parameters()))
```

```python
def accuracy(model, X, y):
    model.eval()
    with torch.no_grad():
        return (model(X).argmax(dim=1) == y).float().mean().item()

def train_classifier(model, X, y, X_v, y_v, epochs=20, lr=3e-3, batch_size=32):
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    loss_fn = nn.CrossEntropyLoss()
    history = {"train": [], "validation": []}
    for epoch in range(epochs):
        model.train()
        order = torch.randperm(len(X))
        for start in range(0, len(X), batch_size):
            batch = order[start:start + batch_size]
            optimizer.zero_grad()
            loss = loss_fn(model(X[batch]), y[batch])
            loss.backward()
            optimizer.step()
        history["train"].append(accuracy(model, X, y))
        history["validation"].append(accuracy(model, X_v, y_v))
    return history

torch.manual_seed(42)
mlp_history = train_classifier(mlp, X_fit, y_fit, X_val, y_val)
print("dense network: validation accuracy =", round(mlp_history["validation"][-1], 4))
```

```python
def plot_history(history, title):
    plt.figure(figsize=(5.5, 3.4))
    plt.plot(range(1, len(history["train"]) + 1), history["train"], marker="o", label="training")
    plt.plot(range(1, len(history["validation"]) + 1), history["validation"],
             marker="o", label="validation")
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.title(title)
    plt.legend()
    plt.tight_layout()
    plt.show()

plot_history(mlp_history, "Dense network: accuracy per epoch")
```

**Inspect (dense network).** The dense network (9,610 weights) reaches 0.977 validation accuracy. Training accuracy rises quickly while validation levels off, so the model fits the training set better than unseen digits.

```python
def make_cnn():
    return nn.Sequential(
        nn.Conv2d(1, 16, kernel_size=3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
        nn.Conv2d(16, 32, kernel_size=3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
        nn.Flatten(),
        nn.Linear(32 * 2 * 2, 64), nn.ReLU(), nn.Dropout(0.2),
        nn.Linear(64, 10),
    )

cnn = make_cnn().to(DEVICE)
print("trainable weights:", sum(p.numel() for p in cnn.parameters()))

sample = X_fit[:1]
for layer in cnn:
    sample = layer(sample)
    if isinstance(layer, (nn.Conv2d, nn.MaxPool2d, nn.Flatten, nn.Linear)):
        print(f"{layer.__class__.__name__:10s} -> {tuple(sample.shape[1:])}")
```

```python
torch.manual_seed(42)
cnn_history = train_classifier(cnn, X_fit, y_fit, X_val, y_val)
print("CNN: validation accuracy =", round(cnn_history["validation"][-1], 4))
plot_history(cnn_history, "CNN: accuracy per epoch")
```

**Inspect (CNN).** The CNN (13,706 weights) reaches 0.9832 validation accuracy, higher than the dense network at 0.977. I kept the 20-epoch model because it was chosen from the validation curve, not from the test file.

```python
test_scores = {
    "logistic on PCA (Lab 4)": 0.9477,
    "dense network": accuracy(mlp, X_test, y_test),
    "CNN": accuracy(cnn, X_test, y_test),
}
for name, value in test_scores.items():
    print(f"{name:26s} test accuracy = {value:.4f}")

plt.figure(figsize=(5, 3.2))
plt.bar(list(test_scores), list(test_scores.values()))
plt.ylim(0.9, 1.0)
plt.ylabel("Official test accuracy")
plt.title("Three models, one test file")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

**Inspect (comparison on the official test file).** Logistic on PCA (Lab 4) = 0.9477, dense network = 0.9583, CNN = 0.9655. Accuracy improves at each step, but each gain is small (about 1 point). The CNN helps because its filters reuse the same learned pattern at every position in the image instead of learning each pixel separately.

```python
cnn.eval()
with torch.no_grad():
    test_pred = cnn(X_test).argmax(dim=1)

ConfusionMatrixDisplay.from_predictions(y_test.cpu().numpy(), test_pred.cpu().numpy(), colorbar=False)
plt.title("CNN confusion matrix on the official test file")
plt.tight_layout()
plt.show()

wrong = torch.nonzero(test_pred != y_test).flatten()[:10]
fig, axes = plt.subplots(1, len(wrong), figsize=(10, 1.6))
for ax, i in zip(axes, wrong):
    ax.imshow(X_test[i, 0].cpu(), cmap="gray_r")
    ax.set_title(f"{int(y_test[i])} read as {int(test_pred[i])}", fontsize=8)
    ax.axis("off")
plt.suptitle("Ten of the CNN's mistakes", y=1.1)
plt.show()
```

**Inspect (errors).** The most frequent confusions in the matrix are 7 read as 5 (7 times), 3 read as 8 (6 times), 7 read as 9 (5 times) and 2 read as 8 (4 times). Of the ten mistakes shown, six are 2s read as 8 or 1, and the images are thick, blocky strokes with a heavy bottom, so an 8x8 grid leaves little detail to tell a 2 from an 8. Many look genuinely ambiguous at this resolution, but the repeated 2-as-8 and 3-as-8 pattern suggests the model leans towards 8 when the shape is unclear, which is a mild systematic bias rather than random noise. Only a few mistakes are shown here, so I would not generalise beyond this.

```python
filters = cnn[0].weight.detach().cpu()[:, 0]
fig, axes = plt.subplots(2, 8, figsize=(8, 2.4))
for ax, kernel in zip(axes.flat, filters):
    ax.imshow(kernel, cmap="coolwarm", vmin=-filters.abs().max(), vmax=filters.abs().max())
    ax.axis("off")
plt.suptitle("The sixteen 3x3 filters of the first convolutional layer", y=1.02)
plt.show()
```

**Inspect (filters).** The sixteen 3x3 first-layer filters show bright and dark contrasts in different orientations. This suggests the network learned suggests the network may have learned edge-like detectors. through loss minimisation, without us hand-specifying edges.

```python
weights_path = MODELS / "digits_cnn.pt"
torch.save(cnn.state_dict(), weights_path)

card = {
    "task": "classify 8x8 handwritten digits (0-9)",
    "input": "tensor of shape (n, 1, 8, 8), pixel values divided by 16",
    "architecture": "make_cnn() in Lab 7: conv16-pool-conv32-pool-dense64-dense10",
    "training": {"epochs": 20, "learning_rate": 3e-3, "batch_size": 32, "seed": 42},
    "official_test_accuracy": round(test_scores["CNN"], 4),
    "torch_version": torch.__version__,
    "trained_on": str(DEVICE),
}
(MODELS / "digits_cnn_card.json").write_text(json.dumps(card, indent=2))

restored = make_cnn().to(DEVICE)
restored.load_state_dict(torch.load(weights_path, map_location=DEVICE, weights_only=True))
restored.eval()
with torch.no_grad():
    same = torch.equal(restored(X_test).argmax(dim=1), test_pred)

size_kb = weights_path.stat().st_size / 1024
print(f"saved {weights_path.name} ({size_kb:.0f} KB); reload reproduces predictions: {same}")
```

**Part D.** The CNN weights and model card are saved in `models/`. The reloaded model reproduced the in-memory predictions (`True`), so the saved file can be trusted. The card records the architecture, training settings, torch version and test accuracy, because the weights alone cannot be reloaded without `make_cnn()`.

```python
results = {}
for ep in (5, 40):
    torch.manual_seed(42)
    m = make_cnn().to(DEVICE)
    results[ep] = train_classifier(m, X_fit, y_fit, X_val, y_val, epochs=ep)
    plot_history(results[ep], f"CNN: {ep} epochs")
    print(ep, "epochs -> validation:", round(results[ep]["validation"][-1], 4))
```

**Try it: 5 vs 40 epochs (validation only).** 5 epochs gives 0.955 validation and 40 epochs gives 0.977, while the 20-epoch CNN gave 0.9832. At 40 epochs training accuracy is almost 1.0 but validation stays around 0.98 and ends slightly lower, so the extra epochs add memorisation, not accuracy. 5 epochs is too few (validation still low), so I keep 20 epochs as the validated choice. Note that the epoch-to-epoch validation values wobble by about 0.01, so small differences should not be over-read.

With a single seed this is within normal run-to-run variation, so the second block gives at most a small gain here. I would need several seeds to say it is a real improvement.
